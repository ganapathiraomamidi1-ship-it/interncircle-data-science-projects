# Project 02 — Titanic Data Visualization

**Goal:** use clear histograms, bar charts, box plots, scatter plots, and a correlation heatmap to communicate patterns in the public seaborn-data Titanic sample.

The figures are saved under `charts/` as standalone PNG files. Chart captions describe observed associations and do not make causal claims.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

candidates = [Path("../data/titanic.csv"), Path("data/titanic.csv"), Path("../../data/titanic.csv")]
DATA_PATH = next((path for path in candidates if path.exists()), None)
if DATA_PATH is None:
    raise FileNotFoundError("Place the included titanic.csv in the portfolio data folder.")
df = pd.read_csv(DATA_PATH)
OUT = Path("charts")
OUT.mkdir(exist_ok=True)
sns.set_theme(style="whitegrid", context="notebook")
print(f"Rows loaded: {len(df)}")
display(df.head())
```

## 1. Passenger age distribution

```python
fig, ax = plt.subplots(figsize=(8, 4.8))
sns.histplot(data=df, x="age", bins=24, color="#2878B5", edgecolor="white", ax=ax)
ax.set(title="Passenger ages are concentrated among younger adults", xlabel="Age (years)", ylabel="Passenger count")
fig.tight_layout()
fig.savefig(OUT / "01_age_distribution.png", dpi=180, bbox_inches="tight")
plt.show()
```

## 2. Survival rates by sex

```python
fig, ax = plt.subplots(figsize=(7, 4.8))
sns.barplot(data=df, x="sex", y="survived", order=["female", "male"], errorbar=None,
            palette={"female": "#D95F76", "male": "#2878B5"}, ax=ax)
ax.set(title="Recorded survival was higher among women", xlabel="Sex", ylabel="Share recorded as survived", ylim=(0, 1))
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda value, _: f"{value:.0%}"))
fig.tight_layout()
fig.savefig(OUT / "02_survival_by_sex.png", dpi=180, bbox_inches="tight")
plt.show()
```

## 3. Fares by passenger class

```python
fig, ax = plt.subplots(figsize=(8, 4.8))
sns.boxplot(data=df, x="class", y="fare", order=["First", "Second", "Third"],
            showfliers=False, color="#82B6A5", ax=ax)
ax.set(title="Typical fares differ substantially across passenger classes", xlabel="Passenger class", ylabel="Fare (dataset currency units)")
ax.text(0.01, -0.24, "Outlier points are hidden for readability; all rows remain in the analysis.", transform=ax.transAxes, fontsize=9)
fig.tight_layout()
fig.savefig(OUT / "03_fare_by_class.png", dpi=180, bbox_inches="tight")
plt.show()
```

## 4. Age, fare, and survival

```python
plot_df = df.dropna(subset=["age", "fare"])
fig, ax = plt.subplots(figsize=(8, 5.2))
sns.scatterplot(data=plot_df, x="age", y="fare", hue="alive", hue_order=["no", "yes"],
                alpha=0.58, s=36, palette={"no": "#4B5563", "yes": "#D95F76"}, ax=ax)
ax.set(title="Age and fare overlap across both recorded outcomes", xlabel="Age (years)", ylabel="Fare (dataset currency units)")
ax.legend(title="Recorded outcome", frameon=True)
fig.tight_layout()
fig.savefig(OUT / "04_age_fare_survival.png", dpi=180, bbox_inches="tight")
plt.show()
```

## 5. Numeric correlation heatmap

```python
numeric = df[["survived", "pclass", "age", "sibsp", "parch", "fare"]].corr()
fig, ax = plt.subplots(figsize=(7, 5.8))
sns.heatmap(numeric, annot=True, fmt=".2f", cmap="vlag", center=0, square=True,
            linewidths=0.5, cbar_kws={"label": "Pearson correlation"}, ax=ax)
ax.set(title="Numeric associations are modest and should not be read as causes")
fig.tight_layout()
fig.savefig(OUT / "05_correlation_heatmap.png", dpi=180, bbox_inches="tight")
plt.show()
```

## Reading the figures

- The bar chart compares group survival proportions, while the heatmap summarizes pairwise numeric associations.
- The scatter plot shows substantial overlap, so a single feature does not separate the outcomes cleanly.
- The fare box plot hides outlier points only to keep the central distributions legible; the source rows are retained.
- These are descriptive charts of a historical dataset, not evidence that any one feature caused survival.
